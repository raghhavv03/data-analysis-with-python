```python
import pandas as pd
```

```python
# Q1.
# A.
df = pd.read_csv("../data/iris.csv")
print(df.head(10))
print(df.shape)
print(df.dtypes)
print(df.describe())
```

```python
# B.
print(df[df["petal_length"] > 4.5])
print(df[df["species"] == "Iris-virginica"])
```

```python
# C.
print(df.groupby("species")["sepal_length"].mean())
print(df.groupby("species")["petal_width"].max())
print(df.groupby("species")["sepal_width"].std())
```

```python
# D.
df1 = df.assign(petal_ratio=df["petal_length"] / df["petal_width"])
print(df1.groupby("species")["petal_ratio"].mean())
```

```python
# Q2.
# A.
df = pd.read_csv("../data/titanic_dataset.csv")
print(df[["Name", "Sex", "Age", "Fare", "Survived"]])
```

```python
# B.
print(df[df["Sex"] == "Female"])
print(df.query("Fare > 30"))
```

```python
# C.
print(df.groupby("Pclass")["Survived"].mean())
print(df.groupby("Pclass")["Fare"].mean())
print(df.groupby("Pclass")["Age"].mean())
```
