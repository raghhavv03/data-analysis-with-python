```python
import requests;
import pandas as pd;
```

```python
# list of pokemon to fetch
pokemon_list = ["ditto", "pikachu", "charizard"]

# collect results here
all_pokemon = []

for name in pokemon_list:
    # build url for this pokemon
    url = f"https://pokeapi.co/api/v2/pokemon/{name}"
    
    # send request
    res = requests.get(url)
    
    # convert response to dict
    json_data = res.json()
    
    # extract fields we need
    core = {
        "name": json_data["name"],
        "id": json_data["id"],
        "height": json_data["height"],
        "weight": json_data["weight"],
        "base_experience": json_data["base_experience"],
    }
    
    # add to list
    all_pokemon.append(core)

# convert to dataframe
df = pd.DataFrame(all_pokemon)
df
```
