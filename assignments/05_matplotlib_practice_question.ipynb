```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["Mumbai", "Chennai", "Bengaluru", "Hyderabad"]

rainfall = [[12, 18, 15, 20, 17], [8, 10, 12, 9, 11], [5, 7, 6, 8, 9], [2, 3, 4, 3, 5]]
# Compare rainfall in cities over a week using subplots and loops.
```

```python
fig, ax = plt.subplots(2, 2, figsize=(7, 7))

for i in range(2):
    for j in range(2):
        ax[i][j].plot(days, rainfall[i + j], marker="o")
        ax[i][j].set_title(cities[i + j])
        ax[i][j].grid()

fig.suptitle("Rainfall in different cities over the week")
fig.supylabel("Rainfall")
fig.supxlabel("Days")
fig.tight_layout()
```

```python
# We can also flatten it.
fig, ax = plt.subplots(2, 2, figsize=(7, 7))

ax = ax.flatten()

for i, ax in enumerate(ax):
    ax.plot(days, rainfall[i], marker="o")
    ax.set_title(cities[i])
    ax.grid()

fig.suptitle("Rainfall in different cities over the week")
fig.supylabel("Rainfall")
fig.supxlabel("Days")
fig.tight_layout()
```
