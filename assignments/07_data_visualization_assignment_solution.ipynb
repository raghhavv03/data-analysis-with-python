```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Q1
scores = np.random.normal(loc = 70, scale = 10, size = 200)

plt.hist(scores, edgecolor = "black", bins = 10)
plt.axvline(
    x = 33,
    label = "passing marks",
    color = "black",
    linewidth = 3,
    linestyle = "--"
)
```

```python
# Q2
penguins = sns.load_dataset("penguins")

sns.scatterplot(
    data = penguins,
    x = "body_mass_g",
    y = "flipper_length_mm",
    hue = "species"
)

plt.title("flipper length vs body mass")
```

```python
# Q3
tips = sns.load_dataset("tips")

sns.boxplot(
    data = tips,
    x = "day",
    y = "total_bill",
    hue = "sex"
)

plt.tight_layout()
```

```python
# Q4
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex",
    style="smoker"
)
```

```python
# Q5
tips_pivot = tips.pivot_table(index = "day", columns = "time", values = "tip", aggfunc = "mean", observed = False)

sns.heatmap(
    tips_pivot,
    cmap = "coolwarm",
    annot = True,
)

plt.tight_layout()
```
