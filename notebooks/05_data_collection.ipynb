```python
# using requests
url = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(url)
```

```python
if res.status_code == 200:
    print(res.headers)
    
with open("../data/scraped-data/data_1.html", "w") as f:
    f.write(res.text)
```

```python
# using beautifulsoup
from bs4 import BeautifulSoup

with open("../data/scraped-data/data_1.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
soup
```

```python
# methods
# soup.find("h1") # finds the first occurence
# soup.h3

all_countries = []

all_h3 = soup.find_all("h3")
for h3 in all_h3:
    name = h3.get_text(strip = True)
    population = h3.find_next("div").select_one("span.country-population").get_text(strip = True)
    all_countries.append([name, population])

df = pd.DataFrame(all_countries, columns = ["Name", "Population"])

df.to_csv("../data/cleaned-data/data.csv")
```
