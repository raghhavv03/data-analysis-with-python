```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 7, 11, 14, 19]
plt.plot(x, y) # line plot
plt.show()
```

```python
movies = [
    "Parasite",
    "Nomadland",
    "CODA",
    "Everything Everywhere All at Once",
    "Oppenheimer"
]
years = [2019, 2020, 2021, 2022, 2023]
oscar_revenues = [262.6, 39.5, 2.2, 143.4, 975.8]

plt.plot(years, oscar_revenues)
plt.title("oscar movies revenue in each year")
plt.xlabel("years")
plt.ylabel("revenue (in millions)")
```

```python
movies = [
    "Avengers: Endgame",
    "Tenet",
    "Spider-Man: No Way Home",
    "Top Gun: Maverick",
    "Barbie"
]
non_oscar_revenues = [2798.0, 365.3, 1921.8, 1495.7, 1446.9]

plt.plot(years, oscar_revenues, label = "oscar movies")
plt.plot(years, non_oscar_revenues, label = "non oscar movies")
plt.title("oscar movies vs non oscar movies revenue")
plt.xlabel("years")
plt.ylabel("revenue (in millions)")
plt.legend(loc = "upper right")
```
