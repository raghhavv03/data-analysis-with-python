```python
import json
```

```python
# load the data
def load_data(file):
    with open(file, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("../data/store_data.json")
print(data)
print(type(data))
```

```python
# clean and structure data
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5} # mapping
    unique_user = set()
    cleaned_data = []
    
    for user in data:
        # cleaning ratings
        raw_rating = str(user["rating"]).strip().lower() 
        if raw_rating in text_to_num:
            user["rating"] = float(text_to_num[raw_rating])
        else:
            user["rating"] = float(raw_rating)

        # handling missing values
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # handling duplicate data
        name = user["name"].strip().lower()
        if(name in unique_user):
           continue
        unique_user.add(name)
        cleaned_data.append(user)
    
    return cleaned_data
```

```python
cleaned_data = clean_data(data)
for user in cleaned_data:
    print(user)
```

```python
# get meaningful insights
def get_insights(cleaned_data):
    # calculating average rating
    sum = 0
    for user in cleaned_data:
        sum += user["rating"]
        average = sum / len(cleaned_data)
    print(f'average rating of the prodict is {average}')

    # percentage of user who gave poor rating
    count = 0
    for user in cleaned_data:
        if(user["rating"] < 3):
            count += 1
    percentage = count / len(cleaned_data) * 100
    print(f'percentage of users who gave poor rating is {percentage}%')
```

```python
get_insights(cleaned_data)
```

```python
# recommendation feature
def get_recommendations(cleaned_data):
    recommendations = []
    
    for user in cleaned_data:
        user_recommendation = {}
        user_recommendation["name"] = user["name"]
        if(user["rating"] >= 4):
            user_recommendation["brand"] = "apple"
        else:
            user_recommendation["brand"] = "samsung"
        recommendations.append(user_recommendation)

    return recommendations
```

```python
recommendations = get_recommendations(cleaned_data)
for user in recommendations:
    print(user)
```
