```python
import pandas as pd
```

```python
# Feature engineering.
df = pd.read_csv("../data/raw_data.csv")
df2 = df.copy()
df2["tax"] = df2["income"].apply(
    lambda x: "20%" if x >= 60000 else "10%"
)  # Creates a new feature (tax) using custom logic
gender_map = {"Male": "M", "Female": "F", "Unknown": "U"}  # Mapping dictionary
df2["gender"] = df2["gender"].map(gender_map)  # Converts values using a dictionary
df2 = df2.assign(
    new_income=df2["income"] * 1.1
)  # Creates new columns without directly modifying existing ones
df2["country"] = df2["country"].replace(
    "USA", "United States"
)  # Substitutes specific values in a column
df2
```

```python
# Feature engineering continued.
df2.columns = [
    "ID",
    "Name",
    "Age",
    "Country",
    "Gender",
    "Income",
    "Tax",
    "New_Income",
]  # Rename all columns
df2.rename(columns={"Income": "Salary"})  # Rename Income column
df2.rename(index={1: "First"})  # Rename index label 1 to First
df2["Income"].sort_values()  # Sort Income column values
df2.sort_values("Income")  # Sort DataFrame by Income
sorted_df = df2.sort_values(
    ["Income", "Age"], ascending=False
)  # Sort by Income and Age in descending order
# Sorted_df.sort_index() # Sort DataFrame by index.
sorted_df = sorted_df.reset_index()  # Reset index and keep old index as a column
# Sorted_df = sorted_df.reset_index(drop = True) # Reset index and drop old index
sorted_df["Ranking"] = sorted_df["Income"].rank(
    ascending=False, method="dense"
)  # Create ranking based on Income
sorted_df = sorted_df[
    ["ID", "Name", "Age", "Country", "Gender", "Income", "New_Income", "Tax"]
]  # Reorder columns
sorted_df
```

```python
# Grouping and aggregation.
df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()
df.groupby("gender")["income"].mean()

# Agg function.
df.groupby("gender")["income"].agg(["mean", "min", "max"])
df.groupby("gender")["income"].aggregate(
    avg_salary="mean", min_salary="min", max_salary="max"
)
df.groupby("country").agg({"income": "max", "age": "mean"})
df.groupby("country").agg(max_salary=("income", "max"), avg_age=("age", "mean"))
```

```python
# Melt and pivot.
df = pd.DataFrame(
    {
        "country": ["usa", "usa", "india", "india"],
        "year": [2020, 2021, 2020, 2021],
        "sales": [100, 120, 90, 110],
        "profit": [20, 25, 18, 22],
    }
)

# Melt.
melted_df = df.melt(
    id_vars=["country", "year"],
    value_vars=["sales", "profit"],
    var_name="metrics",
    value_name="value",
)  # Data converted to long format

# Pivot.
original_df = melted_df.pivot(
    index=["country", "year"],  # Should be a unique value
    columns="metrics",
    values="value",
)
original_df
```

```python
# Merging and joining data.
df_customers = pd.DataFrame(
    {"customer_id": [1, 2, 3, 4], "name": ["adam", "bob", "charlie", "dave"]}
)

df_orders = pd.DataFrame(
    {
        "order_id": [101, 102, 103, 104],
        "customer_id": [2, 1, 4, 5],
        "amount": [250, 120, 300, 180],
    }
)

# Merging.
pd.merge(df_customers, df_orders, on="customer_id")  # Inner join
pd.merge(df_customers, df_orders, on="customer_id", how="left")  # Left join
pd.merge(df_customers, df_orders, on="customer_id", how="right")  # Right join
pd.merge(df_customers, df_orders, on="customer_id", how="outer")  # Outer join
```

```python
# Data concatenation.
df1 = pd.DataFrame({"id": [1, 2, 3], "name": ["adam", "eve", "bob"]})

df2 = pd.DataFrame({"id": [4, 5, 6], "name": ["charlie", "fester", "gillian"]})

pd.concat([df1, df2])  # Row wise
pd.concat([df1, df2], ignore_index=True)  # Assign new labels
pd.concat([df1, df2], axis=1)  # Column wise
```
