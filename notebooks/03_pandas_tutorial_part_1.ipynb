```python
import pandas as pd
```

```python
student_info = {
    "name": ["bob", "alice", "mark"],
    "cgpa": [9.5, 8.75, 8]
}

df = pd.DataFrame(student_info)

print(df)
```

```python
# series
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))

# index based access
print(s[0])
print(s[1])
```

```python
# labelled based access
s = pd.Series([23, 24, 25], index = ["adam", "eve", "bob"])
print(s)
print(s["adam"])
print(s["bob"])
print(s.index)
```

```python
# series properties
# vectorised operations
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
print(s1 + s2)

# mutable values, immutable size
s1[0] = 100
changed_s1 = s1.drop(0)
print(s1) # values are intact
print(changed_s1)
```

```python
# data frame
# creating data frames
# dictionary of lists
student_info = {
    "name": ["bob", "alice", "mark"],
    "age": [25, 29, 31],
    "cgpa": [9.5, 8.75, 8]
}
df = pd.DataFrame(student_info)
print(df)

print(df.index)
print(df.columns)
```

```python
# list of lists
age_info = [["bob", 25], ["charlie", 30], ["mark", 18]]
df = pd.DataFrame(age_info, columns = ["name", "age"])
print(df)
```

```python
# using np arrays
import numpy as np
np_arr = np.array([[1, 2, 3], [4, 5, 6]])
df = pd.DataFrame(np_arr, columns = ["a", "b", "c"])
print(df)                
```
