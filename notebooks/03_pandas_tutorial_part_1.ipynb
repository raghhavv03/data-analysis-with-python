```python
# Aqi data set.
df = pd.read_csv("../data/global_air_quality.csv")
df
```

```python
# Methods on air quality data.
print(df.head())
print(df.describe())
```

```python
# Indexing and selecting data.
# Selecting columns.
df["city"]
df[["city", "aqi"]]  # Case sensitive values

# Selecting rows.
# Label based.
df.loc[0]
df.loc[0:2]  # End idx is inclusive

# Index based.
df.iloc[0]
df.iloc[0:2]  # End idx is exclusive

# Select rows and columns.
df.loc[0, "aqi"]  # Returns a single (scalar) value
df.loc[0, ["city", "aqi"]]  # Returns multiple columns for row 0
df.loc[
    0:2, ["city", "aqi"]
]  # Returns rows 0 to 2 (inclusive) and selected columns (DataFrame)
df.columns  # Shows column names, not numerical indices
df.iloc[
    0:3, [2, 11]
]  # Selects rows 0, 1, 2 and columns at numerical positions 2 and 11
df.iloc[0:3, 2:5]

# Select singular scalar values.
df.at[0, "city"]
df.iat[0, 2]

# View.
cities = df[
    "city"
]  # Creates a view/reference to the original column, not an independent copy
# Cities[0] = "mumbai" # Modifying this may also modify df["city"] because cities is linked to the original DataFrame.
cities = df["city"].copy()  # Create a copy when you plan to modify the data
cities[0] = "Mumbai"
cities
df["city"]  # No change in the original data
```

```python
# Filtering data.
df[(df["aqi"] > 100)]
df[(df["aqi"] > 100) & (df["temperature"] > 30)]
df[(df["aqi"] > 100) & (df["temperature"] < 30)]

# Chaining.
aqi_data = df[(df["aqi"] > 100) & (df["temperature"] > 30)][
    ["city", "aqi", "temperature"]
]

# Labels are preserved.
aqi_data
aqi_data.iloc[0]  # First row by position
aqi_data.loc[6]  # Row with label 6
```

```python
# Filter data using query() method.
df.query("aqi > 100 & temperature > 30")[["city", "aqi"]]
aqi_val = 100
df.query("aqi > @aqi_val & temperature > 30")[["city", "aqi"]]  # Using python variables

# Returns a copy.
# For simplicity, we can always use the copy() method.
aqi_data = df.query("aqi > @aqi_val & temperature > 30")[["city", "aqi"]].copy()
```

```python
# Cleaning data.
# Handling missing values.
df = pd.read_csv("../data/raw_data.csv")
df.isnull()  # Check missing values
df.isnull().sum()  # Count missing values per column
df.dropna()  # Drop rows with missing values
df.dropna(axis=1)  # Drop columns with missing values
df.fillna(0)  # Fill missing values with 0
age_mean = df["age"].mean()  # Calculate mean age
df["age"].fillna(age_mean)  # Returns filled series, does not modify df
cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)  # Update age column
df.ffill()  # Forward fill missing values
df.bfill()  # Backward fill missing values
```

```python
# Handling duplicate values.
df.duplicated()  # Check duplicate rows
df["country"].duplicated()  # Check duplicates in country column
df[["country", "gender"]].duplicated()  # Check duplicates based on country and gender
df.drop_duplicates()  # Return DataFrame without duplicates
# Df.drop_duplicates(inplace=True) # Remove duplicates from original dataframe
```

```python
# Handling data types.
df.dtypes  # View datatype of each column
df2 = df.copy()
df2.fillna(0, inplace=True)
df2["age"] = df2["age"].astype("int64")  # Convert age column to int64
df2.dtypes
date_str = pd.Series(["03-03-2001"])  # Create Series of date strings
print(date_str.dtype)  # Datatype is object
date_str = pd.Series([pd.to_datetime("03-03-2001")])  # Convert string to datetime
print(date_str.dtype)
```

```python
# Handling strings.
df["gender"].str.lower()  # Convert text to lowercase
df["gender"].str.upper()  # Convert text to uppercase
df["gender"].str.capitalize()  # Capitalise first letter
df["name"].str.strip()  # Remove leading and trailing spaces
df["name"].str.split(" ")  # Split names into list using space as separator
df["country"].str.contains(
    "India", case=False
)  # Check if country contains 'India' (case insensitive)
```
