```python
# series
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))

# index based access
print(s[0])
print(s[1])
```

```python
# labelled based access
s = pd.Series([23, 24, 25], index = ["adam", "eve", "bob"])
print(s)
print(s["adam"])
print(s["bob"])
print(s.index)
```

```python
# series properties
# vectorised operations
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
print(s1 + s2)

# mutable values, immutable size
s1[0] = 100
changed_s1 = s1.drop(0)
print(s1) # values are intact
print(changed_s1)
```

```python
# aqi data set
df = pd.read_csv("../data/global_air_quality.csv")
df
```

```python
# methods on air quality data
print(df.head()) 
print(df.describe()) 
```

```python
# indexing and selecting data
# selecting columns
df["city"]
df[["city", "aqi"]] # case sensitive values

# selecting rows
# label based
df.loc[0]
df.loc[0:2] # end idx is inclusive

# index based
df.iloc[0]
df.iloc[0:2] # end idx is exclusive

# select rows and columns
df.loc[0, "aqi"] # returns a single (scalar) value
df.loc[0, ["city", "aqi"]] # returns multiple columns for row 0 
df.loc[0:2, ["city", "aqi"]] # returns rows 0 to 2 (inclusive) and selected columns (dataframe)
df.columns # shows column names, not numerical indices
df.iloc[0:3, [2, 11]] # selects rows 0, 1, 2 and columns at numerical positions 2 and 11
df.iloc[0:3, 2:5]

# select singular scalar values
df.at[0, "city"]
df.iat[0, 2]

# view
cities = df["city"] # creates a view/reference to the original column, not an independent copy
# cities[0] = "mumbai" # modifying this may also modify df["city"] because cities is linked to the original dataframe
cities = df["city"].copy() # create a copy when you plan to modify the data
cities[0] = "mumbai"
cities 
df["city"] # no change in the original data
```

```python
# filtering data
df[(df["aqi"] > 100)]
df[(df["aqi"] > 100) & (df["temperature"] > 30)]
df[(df["aqi"] > 100) & (df["temperature"] < 30)]

# chaining
aqi_data = df[(df["aqi"] > 100) & (df["temperature"] > 30)][["city", "aqi", "temperature"]]

# labels are preserved
aqi_data
aqi_data.iloc[0] # first row by position
aqi_data.loc[6] # row with label 6
```

```python
# filter data using query() method
df.query("aqi > 100 & temperature > 30")[["city", "aqi"]]
aqi_val = 100
df.query("aqi > @aqi_val & temperature > 30")[["city", "aqi"]] # using python variables
# returns a copy
# for simplicity, we can always use the copy() method
aqi_data = df.query("aqi > @aqi_val & temperature > 30")[["city", "aqi"]].copy()
```
